```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Load data
df = pd.read_csv("online_food_delivery_cleaned.csv")

# Set up a multi-plot dashboard layout (2 rows, 2 columns)
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle(
    "Online Food Delivery - Data Visualization Dashboard",
    fontsize=16,
    fontweight="bold",
)

# 1. Bar Chart: Customer Count by Occupation
sns.countplot(
    data=df, x="Occupation", ax=axes[0, 0], palette="Set2", hue="Occupation"
)
axes[0, 0].set_title("Customer Distribution by Occupation")
axes[0, 0].tick_params(axis="x", rotation=15)

# 2. Countplot: Feedback Breakdown by Gender
sns.countplot(
    data=df,
    x="Gender",
    hue="Feedback",
    ax=axes[0, 1],
    palette="pastel",
    legend=True,
)
axes[0, 1].set_title("Feedback Count by Gender")

# 3. Histogram: Customer Age Distribution
sns.histplot(df["Age"], bins=10, kde=True, ax=axes[1, 0], color="purple")
axes[1, 0].set_title("Customer Age Distribution")

# 4. Countplot: Monthly Income Levels
sns.countplot(
    data=df,
    y="Monthly Income",
    ax=axes[1, 1],
    order=df["Monthly Income"].value_counts().index,
    palette="coolwarm",
    hue="Monthly Income",
)
axes[1, 1].set_title("Monthly Income Levels")

# Adjust layout and display
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```
